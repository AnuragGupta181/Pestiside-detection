```python
import os

labels_dir = "./agri_data/labels"

for filetxt in os.listdir(labels_dir):
    if filetxt.endswith(".txt"):
        path = os.path.join(labels_dir, filetxt)
        with open(path, "r") as f:
            lines = f.readlines()
        for i, line in enumerate(lines):
            parts = line.strip().split()
            if len(parts) != 5:
                print(f"⚠️ Invalid format in {filetxt}, line {i+1}: {line.strip()}")
            if parts[0] not in ["0","1"]:
                print(f"⚠️ Invalid class ID in {filetxt}, line {i+1}: {line.strip()}")
```
